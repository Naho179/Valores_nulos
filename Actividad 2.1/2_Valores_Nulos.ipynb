**Métodos de Sustitución de Valores Nulos**

```python
#Realizamos una copia del dataframe
data3=data2.copy()
```

```python
#Primer método de sustitución de valores nulos
#Sustituir valores nulos con promedio o media
data3["Superficie"]=data3["Superficie"].fillna(round(data2["Superficie"].mean(),1))
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Segundo método de sustitución de valores nulos
#Sustituir valores nulos con mediana
data3["Precio Venta"]=data3["Precio Venta"].fillna(round(data2["Precio Venta"].median(),1))
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Tercer método de sustitución de valores nulos
#Sustituir valores nulos por un valor numérico en  concreto 
data3["Fecha Alta"]= data3["Fecha Alta"].fillna(10) 
data3
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Cuarto método de sustitución de valores nulos
#Sustituir valores nulos por un string en  concreto 
data3["Fecha Venta"] =data3["Fecha Venta"].fillna("2030-06-19 00:00:00") 
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Quinto método de sustitución de valores nulos
#Sustituir valores nulos por valores no nulos hacia adelante  "forward fill" ("ffill")
#Filtro por columnas
data3["Tipo"] =data3["Tipo"].ffill() ##ff= forwardfill recorre columnas y cuando encuentra un valor nulo, lo llena con el valor anterior
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Sexto método de sustitución de valores nulos
#Sustituir valores nulos por valores no nulos hacia atrás backward fill" ("bfill")
#Filtro por columnas
data3["Operación"] =data3["Operación"].bfill() ##valor previo del recorrido
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por valores no nulos hacia atrás backward fill" ("bfill")
data3["Provincia"] =data3["Provincia"].bfill() 
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por valores no nulos hacia adelante forward fill" ("ffill")
#Filtro por columnas
data3["Vendedor"] =data3["Vendedor"].ffill() 
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Corroboramos valores nulos en todo el dataframe
valores_nulos=data3.isnull().sum().sum()
valores_nulos
```

```python
##LIMPIEZA DE NULOS EN CSV 

##ACTIVIDAD 2.2
df_nulos = pd.read_csv('Copia de Citas_Digital_Filtrado.csv')
df_nulos.head(5)
```

```python
df_nulos.columns
```

```python
df_nulos.isnull().sum()
```

```python
#Para esta columna decidí rellenas los valores nulos por el texto "Se ignora" ya que 
#al albergar nombres, son datos que no podemos imputar sin generar sesgos
df_nulos['BDC '] = df_nulos['BDC '].fillna("Se ignora")
```

```python
#Para columna de fecha, cree una nueva columna con el nombre de "Mes" 
#y por medio de rangos asigné la fecha correspondiente a los registros, incluyendo mes y año
#correspondiente.

#Posteriormente reemplacé la columna original de "Fecha" por esta nueva columna.


df_nulos["Mes"] = None
```

```python
df_nulos.loc[0:30, "Mes"] = "2025-04"
df_nulos.loc[34:58, "Mes"] = "2025-05"
df_nulos.loc[61:86, "Mes"] = "2025-06"
df_nulos.loc[88:120, "Mes"] = "2025-07"
df_nulos.loc[122:166, "Mes"] = "2025-08"
df_nulos.loc[168:230, "Mes"] = "2025-09"
df_nulos.loc[232:290, "Mes"] = "2025-10"
df_nulos.loc[292:343, "Mes"] = "2025-11"
df_nulos.loc[345:391, "Mes"] = "2025-12"

df_nulos.loc[393:428, "Mes"] = "2026-01"
df_nulos.loc[430:460, "Mes"] = "2026-02"
df_nulos.loc[462:477, "Mes"] = "2026-03"
df_nulos.loc[479:533, "Mes"] = "2026-04"
df_nulos.loc[540:598, "Mes"] = "2026-05"

fechas_reales = pd.to_datetime(df_nulos["Fecha"], errors="coerce")

df_nulos.loc[600:672, "Mes"] = (
    fechas_reales.loc[600:672]
    .dt.to_period("M")
    .astype("string")
)
```

```python
df_nulos[["BDC ", "Fecha", "Nombre Cliente ", "Mes"]].tail(80)
```

```python
df_nulos["Mes"].value_counts().sort_index()
```

```python
df_nulos["Mes"].isnull().sum()
```

```python
# 3. Ver únicamente las filas que no tienen mes
df_nulos[df_nulos["Mes"].isnull()]
```

```python
## Aquí comprobé que los registros nulos que había en mi columna de mes se debía a 
# filas basura (la mayoría, filas de separación entre meses) que podían ser eliminadas sin problema.
df_nulos[df_nulos["Mes"].isnull()][["BDC ", "Fecha", "Nombre Cliente ", "Estatus de Lead"]]
```

```python
# Eliminación de las filas con datos nulos en "Mes"
df_limpio = df_nulos.dropna(subset=["Mes"]).copy()
```

```python
df_limpio["Mes"].isnull().sum()
```

```python
# Eliminación de columna original "Fecha"
df_limpio = df_limpio.drop(columns=["Fecha"])

# Renombre, cambiamos Mes por Fecha
df_limpio = df_limpio.rename(columns={"Mes": "Fecha"})
```

```python
#Nueva columna en nueva posición
fecha = df_limpio.pop("Fecha")
df_limpio.insert(1, "Fecha", fecha)
```

```python
df_limpio.isnull().sum()

## Al eliminar las filas basura mencionadas anteriromente, también se eliminaron los nulos 
# de PDM y SDC
```

```python
df_limpio.columns
```

```python
# Esta columna la consideré como un caso similar a la columna de "BDC ", al ser columna cualitativa 
# creo importante no sesgar la información con moda o algo similar, ya que supongo que este estatus tendrá más factores o info relevante asociada.
df_limpio["Estatus de Lead"] = (df_limpio["Estatus de Lead"].fillna("No hay registro"))
```

```python
df_limpio["Potencial de compra"] = pd.to_numeric(df_limpio["Potencial de compra"],errors="coerce")
```

```python
# Inicialmente había llenado las celdas vacía con NaN, debido a que igual que en el caso
# de "Estatus de Lead" considero que esta columna debe estar asociada a más factores y podría ser sesgada fácilmente
# sin embargo como seguíann apareciendo como valores nulos, al ser datos cuantitativos, 
# decidí implmentar la mediana, un dato más representativo que la media.

mediana_potencial = df_limpio["Potencial de compra"].median()

df_limpio["Potencial de compra"] = (df_limpio["Potencial de compra"].fillna(mediana_potencial))
```

```python
mediana_potencial
```

```python
df_limpio.tail()
```

```python
df_limpio[df_limpio["Venta"].isnull()]
```

```python
## Al ser una variable binaria, decidí reemplazar los pocos valores nulos con la moda.
moda_venta = df_limpio["Venta"].mode()[0]
df_limpio["Venta"] = df_limpio["Venta"].fillna(moda_venta)
```

```python
## Igual que en "BDC", al ser NOMBRES de los asesores de la empresa, 
# decidí llenar las celdas con "Se ignora" para no generar mala info.
df_limpio['Asesor Asignado'] = df_limpio['Asesor Asignado'].fillna("Se ignora")
```

```python
df_limpio.isnull().sum()
```

```python
## Estas últimas columnas decidí tratarlas de la misma manera, ya que al ser texto preferí llenar 
# los valores nulos con 'No hay reggistro'.

texto = ['¿Por que no ha visitado la agencia?','¿Por que no hay solicitud de credito?','¿Por que no hubo prueba de manejo ?','¿Por que no hubo proceso wow?','¿Por que se asigno antes de vistar piso?']
df_limpio[texto] = df_limpio[texto].fillna('No hay registro')
```
