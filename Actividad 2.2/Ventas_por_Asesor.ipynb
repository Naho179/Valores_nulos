```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("Ventas_por_Asesor.xlsx", header=1)
```

```python
columnas_ventas = df.columns[1:31]

df[columnas_ventas] = df[columnas_ventas].replace(r"^\s*$", np.nan, regex=True)
df[columnas_ventas] = df[columnas_ventas].fillna(0)

## Se identificaron espacios vacíos que no estaban registrados como NaN, 
# por lo que se convirtieron a valores nulos y después se reemplazaron por 0 
# en las columnas de ventas
```

```python
## Renombré la primera columna como Nombre vendedor y eliminé la fila que repetía 
# ese encabezado, para dejar la estructura de la base organizada.
df = df.rename(columns={"Unnamed: 0": "Nombre vendedor"})
df = df.iloc[1:].reset_index(drop=True)
```

```python
## Comprobamos solamente que no haya valores nulos en nuestras columnas asociadas a las ventas
# de asesores
df.iloc[:, 1:31].isnull().sum()
```

```python
## Como aclaración, los nulos de las columnas de resumen (suma de ventas y ranking) se dejaron sin cambios 
# porque no representan ventas faltantes, sino un error de valor derivado de la
# ejecución de ceirta fórmula en excel.
```

```python
df.to_csv("Ventas_por_Asesor_sin_nulos.csv")
```
