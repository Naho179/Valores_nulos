```python
## Decidí eliminar directamente las columnas: Unnamed 17 y 18 debido a que no tienen un solo registro.
# Eliminé: unnamed 19 porque solo tiene 1 valor no nulo
# Eliminé unnamed 16 porque no tenía info clara ni un nombre acorde al contenido de la columna.

df = df.drop(columns=["Unnamed: 16","Unnamed: 17", "Unnamed: 18","Unnamed: 19"])
```

```python
## Ver dónde están los nulos
df[df["Fecha Origen"].isnull()]
```

```python
# Rellené fechas faltantes con la fecha anterior, ya que sólo eran 2 registros faltantes 
df["Fecha Origen"] = df["Fecha Origen"].ffill()
```

```python
df[df["Nombre"].isnull()]
```

```python
## Se elimina la fila basura (fila de separación) comprobada en el código anterior, 
# con lo que se eliminan los valores nulos de Nombre, Producto, Asesor y Origen.
df = df.drop(index=194)
df = df.reset_index(drop=True)
```

```python
df.isnull().sum()
```

```python
df[df["Seguimiento"].isnull()]
```

```python
df[df["Seguimiento"].isnull()]
## Se conservaron los valores faltantes en Seguimiento, Cita efectiva, PDM y SDC porque 
# no había información suficiente para imputarlos sin inventar datos. 
```

```python
# Se usó Int64 para mantener los valores 0-1 y representar los faltantes como <NA>
df["Seguimiento"] = df["Seguimiento"].astype("Int64")
df["Cita efetiva "] = df["Cita efetiva "].astype("Int64")
df["PDM"] = df["PDM"].astype("Int64")
df["SDC"] = df["SDC"].astype("Int64")
```

```python
df.isnull().sum()
```

```python
df.columns
```

```python
## Los valores faltantes en las columnas de texto se reemplazaron por No hay registro, 
# ya que no hay manera de imputarlos.

texto = ['Estatus','¿Por que no hay seguimiento?','¿Por que no ha visitado la agencia?','¿Por que no hubo prueba de manejo ?','¿Por que no hubo solicitud de credito?','¿Por que no se ha formalizado la compra?']
df[texto] = df[texto].fillna('No hay registro')
```

```python
df.isnull().sum()
```

```python
## A manera de resumen y para esclarecer  la razón por la que algunos valores nulos 
# siguen apareciendo, es porque los valores faltantes se conservaron como <NA> para 
# mantener las variables binarias en formato numérico y evitar mezclar string con números.
```
